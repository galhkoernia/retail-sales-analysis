```python
from pathlib import Path

import pandas as pd
```

```python
DATA_DIR = Path("../data")
PROCESSED_DATA_PATH = DATA_DIR / "processed" / "sales_data_cleaned.csv"

pd.set_option("display.max_columns", None)
pd.set_option("display.float_format", "{:,.2f}".format)
```

```python
sales_df = pd.read_csv(
    PROCESSED_DATA_PATH,
    parse_dates=["ORDERDATE"]
)

sales_df.head()
```

```python
# Business Question 1
# Bagaimana total penjualan setiap tahun?

yearly_sales = (
    sales_df
    .groupby("YEAR_ID", as_index=False)
    .agg(
        Total_Sales=("SALES", "sum")
    )
    .sort_values("YEAR_ID")
)

yearly_sales

best_year = yearly_sales.loc[
    yearly_sales["Total_Sales"].idxmax()
]

best_year

```

```python
# Business Question 2
# Negara dengan penjualan Classic Cars terbesar pada tahun 2004

classic_2004 = sales_df[
    (sales_df["YEAR_ID"] == 2004) &
    (sales_df["PRODUCTLINE"] == "Classic Cars")
]

country_sales = (
    classic_2004
    .groupby("COUNTRY", as_index=False)
    .agg(
        Total_Sales=("SALES", "sum")
    )
    .sort_values(
        by="Total_Sales",
        ascending=False
    )
)

country_sales

top_country = country_sales.iloc[0]

top_country
```

```python
# Business Question 3
# Product Line dengan Quantity Ordered terbesar pada periode September 2003 hingga Maret 2004
filtered_df = sales_df[
    (sales_df["ORDERDATE"] >= "2003-09-01") &
    (sales_df["ORDERDATE"] <= "2004-03-31")
]

product_quantity = (
    filtered_df
    .groupby("PRODUCTLINE", as_index=False)
    .agg(
        Total_Quantity=("QUANTITYORDERED", "sum")
    )
    .sort_values(
        by="Total_Quantity",
        ascending=False
    )
)

product_quantity

top_product = product_quantity.iloc[0]

top_product
```

```python
# Executive Summary
# Business Question 1
best_year = (
    sales_df.groupby("YEAR_ID", as_index=False)
    .agg(Total_Sales=("SALES", "sum"))
    .sort_values("Total_Sales", ascending=False)
    .iloc[0]
)

# Business Question 2
top_country = (
    sales_df[
        (sales_df["YEAR_ID"] == 2004) &
        (sales_df["PRODUCTLINE"] == "Classic Cars")
    ]
    .groupby("COUNTRY", as_index=False)
    .agg(Total_Sales=("SALES", "sum"))
    .sort_values("Total_Sales", ascending=False)
    .iloc[0]
)

# Business Question 3
top_product = (
    sales_df[
        (sales_df["ORDERDATE"] >= "2003-09-01") &
        (sales_df["ORDERDATE"] <= "2004-03-31")
    ]
    .groupby("PRODUCTLINE", as_index=False)
    .agg(Total_Quantity=("QUANTITYORDERED", "sum"))
    .sort_values("Total_Quantity", ascending=False)
    .iloc[0]
)

summary = pd.DataFrame({
    "Business Question": [
        "Highest Sales Year",
        "Top Country (Classic Cars - 2004)",
        "Top Product Line (Sep 2003 - Mar 2004)"
    ],
    "Key Result": [
        best_year["YEAR_ID"],
        top_country["COUNTRY"],
        top_product["PRODUCTLINE"]
    ],
    "Metric": [
        f"${best_year['Total_Sales']:,.2f}",
        f"${top_country['Total_Sales']:,.2f}",
        f"{top_product['Total_Quantity']:,} Units"
    ]
})

summary
```
