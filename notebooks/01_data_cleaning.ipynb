```python
from pathlib import Path

import pandas as pd
import chardet
```

```python
# Paths
DATA_DIR = Path("../data")
RAW_DATA_PATH = DATA_DIR / "raw" / "sales_data.csv"
PROCESSED_DATA_PATH = DATA_DIR / "processed" / "sales_data_cleaned.csv"

# Display Options
pd.set_option("display.max_columns", None)
pd.set_option("display.float_format", "{:,.2f}".format)
```

```python
try:
    sales_df = pd.read_csv(
        RAW_DATA_PATH,
        sep="\t",
        decimal=",",
        encoding="cp1252",
        low_memory=False
    )
except Exception:
    sales_df = pd.read_excel(
        RAW_DATA_PATH,
        engine="openpyxl"
    )

sales_df.head()
```

```python
print(f"Number of Rows    : {sales_df.shape[0]:,}")
print(f"Number of Columns : {sales_df.shape[1]}")
```

```python
sales_df.head()
```

```python
sales_df.info()

sales_df.describe(include="all").T
```

```python
missing_values = (
    sales_df
    .isna()
    .sum()
    .to_frame(name="Missing Values")
)

missing_values["Percentage"] = (
    missing_values["Missing Values"]
    / len(sales_df)
    * 100
).round(2)

missing_values.sort_values(
    by="Missing Values",
    ascending=False
)
```

```python
duplicate_count = sales_df.duplicated().sum()

print(f"Duplicate Records : {duplicate_count}")
```

```python
categorical_columns = sales_df.select_dtypes(include="object").columns

for column in categorical_columns:
    print(f"{column:20} : {sales_df[column].nunique()}")
```

```python
# Data Cleaning
# Remove Duplicate

rows_before = len(sales_df)

sales_df = (
    sales_df
    .drop_duplicates()
    .reset_index(drop=True)
)


rows_after = len(sales_df)

print(f"Duplicate Removed : {rows_before - rows_after}")

# Remove Extra Whitespac
text_columns = sales_df.select_dtypes(include="object").columns

sales_df[text_columns] = sales_df[text_columns].apply(
    lambda column: column.str.strip()
)

# Convert Date
sales_df["ORDERDATE"] = pd.to_datetime(
    sales_df["ORDERDATE"],
    format="%m/%d/%Y %H:%M",
    errors="coerce"
)

sales_df["ORDERDATE"].dtype

sales_df["ORDERDATE"].isna().sum()

numeric_columns = [
    "QUANTITYORDERED",
    "PRICEEACH",
    "ORDERLINENUMBER",
    "SALES",
    "QTR_ID",
    "MONTH_ID",
    "YEAR_ID",
    "MSRP"
]

sales_df[numeric_columns].dtypes

categorical_columns = [
    "STATUS",
    "PRODUCTLINE",
    "COUNTRY",
    "TERRITORY",
    "DEALSIZE"
]

for column in categorical_columns:
    print(f"\n{column}")
    print(sales_df[column].sort_values().unique())

sales_df.isna().sum()
```

```python
print(sales_df.shape)

sales_df.head()

```

```python
PROCESSED_DATA_DIR.mkdir(parents=True, exist_ok=True)

sales_df.to_csv(
    PROCESSED_DATA_PATH,
    index=False
)

print(f"Dataset exported to:\n{PROCESSED_DATA_PATH}")
```
